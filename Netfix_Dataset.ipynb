```python
#step 1 Import The Libraries
import pandas as pd
import matplotlib.pyplot as plt

#load The Data
df = pd.read_csv('Netflix_dataset.csv')

#clean Data
df = df.dropna(subset = ['show_id','type','title','director','cast','country','date_added','release_year','rating','duration','listed_in','description'])

type_counts = df['type'].value_counts()
plt.figure(figsize=(3,4))
plt.bar(type_counts.index, type_counts.values,color = ['skyblue','orange'])
plt.title('Number of Movies Vs Tv Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Counts')
plt.tight_layout()
plt.savefig('Movies_vs_tvshows.png')
plt.show()



rating_counts = df['rating'].value_counts()
plt.figure(figsize=(8,6))
plt.pie(rating_counts, labels=rating_counts.index, autopct='%1.1f%%',startangle=90)
plt.title('Percantage of content Rating')
plt.tight_layout()
plt.savefig('RatingofContent.png')
plt.show()

country_counts = df['country'].value_counts()
plt.figure(figsize=(8,6))
plt.barh(country_counts.index,country_counts.values,color=['Red','Grey'])
plt.title('Number of countries ')
plt.xlabel('Countries')
plt.ylabel('Counts')
plt.tight_layout()
plt.savefig('countries.png')
plt.show()

year_counts = df['release_year'].value_counts()
plt.figure(figsize=(8,6))
plt.bar(year_counts.index,year_counts.values,color =['blue','red'])
plt.title('Release Year Comparison ')
plt.xlabel('Years')
plt.ylabel('Counts')
plt.tight_layout()
plt.savefig('Release_Year.png')
plt.show()

year_counts = df['release_year'].value_counts()
country_counts = df['country'].value_counts()
```
