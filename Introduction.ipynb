```python
import matplotlib.pyplot as plt
x = ['Mon', 'Tue', 'Wed', 'Thur', 'Fri']
y = [10, 15, 7, 20, 12]
plt.plot(x,y, color = 'Green', linewidth = '2', linestyle ='--', marker = 'o' ,label = 'Sales This Week')
plt.title('Bakery sales This Week')
plt.xlabel('Day of the week')
plt.ylabel('sales Per Day')

plt.legend(loc= 'upper left', fontsize = '12')
plt.grid(color = 'Yellow', linestyle = ':',linewidth = '2')
plt.xlim(0,5)
plt.ylim(0,20)

plt.show()
```

```python
import matplotlib.pyplot as plt
Regions = ['North','South','East','West']
Revenue = [3000,2000,1500,1000]
plt.pie(Revenue,labels=Regions,autopct='%1.1f%%',colors =['Yellow','Skyblue','Red','coral'])
plt.title('Revenue Contrebutaion By Regions')
plt.show()
```

```python
import matplotlib.pyplot as plt
scores = [45,67,89,56,78,34,46,23,84,43,68,87,37,35,65,38,47,32]
plt.hist(scores,bins=7,color='Red',edgecolor ='black')
plt.xlabel('Score Range')
plt.ylabel('Number of Students')
plt.title('Score Distribution of Students')
plt.show()
```

```python
import matplotlib.pyplot as plt
Hours_study = [1,2,3,4,5,6,7,8]
Exams_Score = [50,55,60,75,77,86,89,93]
plt.scatter(Hours_study,Exams_Score, color='green',marker = '^', label = 'Student Data')
plt.xlabel('Hours Study')
plt.ylabel('Exam Scores')
plt.title('Relation Between Hours Study and Exams Score')
plt.legend()
plt.grid()
plt.show()
```

```python
#plt.subplot(nrows,ncloms,index)
import matplotlib.pyplot as plt
x=[1,2,3,4]
y=[10,20,15,25]
plt.subplot(1,2,1) #1 Row ,1colum, 1st Subplot
plt.plot(x,y,color = "Blue")
plt.title('Line Chart')

plt.subplot(1,2,2) #1 ROW , 2 COLOM, 2ND SUBPLOT
plt.bar(x,y, color = 'Red')
plt.title('Bar Chart ')
plt.tight_layout() #this is For complete the canvas with graph (Automatic Adjustment , overlap Fix)
plt.show()
```

```python
#the Recomnded Way to show in Interviews
#fig, ax = plt.subplot(nrows,ncoloms,figsize=(width ,height))

import matplotlib.pyplot as plt
fig, ax = plt.subplots(1,2, figsize = (10,5))
x=[1,2,3,4]
y=[10,20,15,25]

ax[0].plot(x,y)
ax[0].set_title('Line Chart')

ax[1].bar(x,y)
ax[1].set_title('Bar Chart')
fig.suptitle('Comparision Of Line And Bar Charts')
plt.tight_layout()
plt.savefig('subplot_chart.pdf',dpi = 300 , bbox_inches = "tight")
plt.show()
```
